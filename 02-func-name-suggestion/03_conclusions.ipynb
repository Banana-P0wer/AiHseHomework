# 03. Итоговые выводы

## Постановка задачи

В работе решалась задача автоматической генерации имени Python-функции по её
телу. Для анализа использованы реальные функции из CodeSearchNet, а для
генерации — предобученная модель CodeT5+ без дополнительного обучения.


## Итоговые результаты

| Входные данные | Exact Match | ROUGE-1 | ROUGE-2 | ROUGE-L |
| --- | ---: | ---: | ---: | ---: |
| Только код | 0.143 | 0.384 | 0.198 | 0.381 |
| Код, документация и комментарии | 0.184 | 0.467 | 0.267 | 0.465 |

Добавление документации и комментариев улучшает результат по всем метрикам.


## Основные наблюдения

- CodeT5+ без дополнительного обучения способна предлагать осмысленные имена
  функций на основе исходного кода.
- Текстовая документация помогает модели точнее выразить назначение функции.
- Exact Match является строгой метрикой: разные корректные имена функции могут
  не совпадать посимвольно.


## Ограничения подхода

- Модель не обучалась дополнительно на рассматриваемой задаче.
- Вход ограничен 512 токенами, поэтому длинные функции могут обрезаться.
- Оценка сравнивает предсказание только с одним исходным именем, хотя
  смыслово корректных вариантов может быть несколько.


## Итоговый вывод

В работе подготовлена выборка функций, выполнен их синтаксический разбор и
проведено сравнение двух режимов генерации имён. Документация и комментарии
заметно повышают качество результата, что подтверждает ценность текстового
контекста при анализе исходного кода.
