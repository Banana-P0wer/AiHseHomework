# 03. Final Conclusions

## Problem Statement

This project addressed automatic generation of a Python function name from its
body. Real functions from CodeSearchNet were analyzed, and the pretrained
CodeT5+ model was used for generation without additional training.


## Final Results

| Input data | Exact Match | ROUGE-1 | ROUGE-2 | ROUGE-L |
| --- | ---: | ---: | ---: | ---: |
| Code only | 0.143 | 0.384 | 0.198 | 0.381 |
| Code, documentation, and comments | 0.184 | 0.467 | 0.267 | 0.465 |

Adding documentation and comments improves the result on every metric.


## Main Observations

- Without additional training, CodeT5+ can suggest meaningful function names
  from source code.
- Natural-language documentation helps the model express function intent more
  precisely.
- Exact Match is a strict metric: different correct function names may not
  match character by character.


## Limitations

- The model was not additionally trained for the target task.
- Input is limited to 512 tokens, so long functions may be truncated.
- Evaluation compares a prediction with only one original name, although
  several semantically correct alternatives may exist.


## Final Conclusion

The project prepared a function sample, parsed its syntax, and compared two
function-name generation modes. Documentation and comments noticeably improve
the result, confirming the value of natural-language context in source-code
analysis.
