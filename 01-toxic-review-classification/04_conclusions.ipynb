# 04. Final Conclusions

## Problem Statement

This project addressed binary classification of source-code review comments as
toxic or non-toxic. After cleaning the data, two models were trained and
compared: TF-IDF with logistic regression and RoBERTa.


## Final Results

| Model | Accuracy | Toxic precision | Toxic recall | Toxic F1 |
| --- | ---: | ---: | ---: | ---: |
| TF-IDF + Logistic Regression | 0.912 | 0.894 | 0.927 | 0.910 |
| RoBERTa | 0.925 | 0.897 | 0.954 | 0.924 |

Both models noticeably exceeded the assignment reference level of
`F1 approximately 0.7`. RoBERTa improved toxic-class F1 by `0.014` relative
to the classical model.


## Main Observations

- TF-IDF and logistic regression provide a strong and explainable result:
  terms increasing the probability of toxicity can be inspected directly.
- RoBERTa identifies toxic examples with better recall (`0.954` versus
  `0.927`) because it can account for contextual expressions.
- Words used in programming require context-sensitive interpretation:
  for example, `kill` may refer to a process rather than abuse.


## Limitations

- The training split is imbalanced: toxic examples are substantially fewer.
- The comparison is conducted on a small test split of 228 comments.
- The transformer provides a better score but requires more resources and is
  harder to interpret.


## Final Conclusion

Both approaches successfully address toxic comment classification. Logistic
regression is a strong transparent baseline, while RoBERTa delivers better
overall quality when textual context matters.
