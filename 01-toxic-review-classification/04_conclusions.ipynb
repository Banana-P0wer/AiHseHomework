# 04. Итоговые выводы

## Постановка задачи

В работе решалась задача бинарной классификации комментариев к исходному коду
на токсичные и нетоксичные. После очистки данных были обучены и сравнены две
модели: TF-IDF с логистической регрессией и RoBERTa.


## Итоговые результаты

| Модель | Accuracy | Precision для токсичных | Recall для токсичных | F1 для токсичных |
| --- | ---: | ---: | ---: | ---: |
| TF-IDF + Logistic Regression | 0.912 | 0.894 | 0.927 | 0.910 |
| RoBERTa | 0.925 | 0.897 | 0.954 | 0.924 |

Обе модели заметно превышают ориентир задания `F1 ≈ 0.7`. RoBERTa улучшает
F1 токсичного класса на `0.014` относительно классической модели.


## Основные наблюдения

- TF-IDF и логистическая регрессия обеспечивают сильный и объяснимый результат:
  можно изучать слова, увеличивающие вероятность токсичного класса.
- RoBERTa лучше распознаёт токсичные примеры по полноте (`0.954` против
  `0.927`), поскольку способна учитывать контекст выражений.
- Слова, встречающиеся в программировании, требуют контекстного анализа:
  например, `kill` может относиться к процессу, а не к оскорблению.


## Ограничения подхода

- Обучающая выборка несбалансирована: токсичных примеров заметно меньше.
- Сравнение проведено на небольшой тестовой выборке из 228 комментариев.
- Трансформер даёт более высокий результат, но требует больше ресурсов и
  сложнее интерпретируется.


## Итоговый вывод

Оба подхода успешно решают задачу классификации токсичных комментариев.
Логистическая регрессия является хорошей прозрачной базовой моделью, а
RoBERTa показывает лучшее итоговое качество, когда важен контекст текста.
